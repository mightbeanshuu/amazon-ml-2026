```python
import os, pandas as pd, re
data = None
for root, dirs, files in os.walk('/kaggle/input', followlinks=True):
    if 'train_source1.tsv' in files and root.endswith('/train'):
        data = root[:-6]
        break
norm = lambda s: re.sub(r'[^a-z0-9]', '', str(s).lower())
rd = lambda p: pd.read_csv(p, sep='\t', dtype=str, keep_default_na=False, quoting=3)
tr1 = rd(data + '/train/train_source1.tsv'); te1 = rd(data + '/test/test_source1.tsv')
tr1['k'] = tr1.business_name.map(norm) + '|' + tr1.business_address.map(norm)
te1['k'] = te1.business_name.map(norm) + '|' + te1.business_address.map(norm)
inter = set(tr1.k) & set(te1.k)
print('S1 exact name+addr overlap:', len(inter), 'of', len(te1), '=', round(len(inter)/len(te1), 4))
tr1['kn'] = tr1.business_name.map(norm); te1['kn'] = te1.business_name.map(norm)
print('S1 name-only overlap:', len(set(tr1.kn) & set(te1.kn)), 'of', te1.kn.nunique())
tr2 = rd(data + '/train/train_source2.tsv'); te2 = rd(data + '/test/test_source2.tsv')
tr2['k'] = tr2.business_name.map(norm) + '|' + tr2.business_address.map(norm)
te2['k'] = te2.business_name.map(norm) + '|' + te2.business_address.map(norm)
print('S2 record overlap:', len(set(tr2.k) & set(te2.k)), 'of', len(te2))
gt = rd(data + '/train/train_ground_truth.tsv')
print('gt rows', len(gt))
print('PROBE DONE')
```
